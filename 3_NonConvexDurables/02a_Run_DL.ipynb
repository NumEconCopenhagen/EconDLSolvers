## 2. <a id='toc2_'></a>[1D](#toc0_)

### 2.1. <a id='toc2_1_'></a>[Sigmoid](#toc0_)

```python
device = choose_gpu()
K_time = 120
```

```python
par = {'D':1, 'full':True}
train  = {'K_time':K_time,'budget_shares':False}

model = NonConvexDurablesModelClass(device=device,algoname='DeepVPDDC',train=train, par=par)
model.solve(do_print=True)
```

```python
model.more_simulation_outcomes()
model.euler_errors_DL()
```

```python
model.save('../output/NonConvexDurablesModel_DL_1D_sigmoid.pt')
```
