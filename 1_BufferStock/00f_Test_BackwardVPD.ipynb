## 2. <a id='toc2_'></a>[Settings](#toc0_)

```python
LOAD_SIMULT = False
K_time_simult = 1.0 
K_time = 1.0
```

```python
algoname = 'DeepVPD'
filename = '../output/BufferStockModelVPD_DeepVPD_0D_NNs3.pt'
```

```python
def print_R(model):
    model.simulate_R()
    R = model.sim.R.item()
    print(f'{R = :12.8f}')
```

## 3. <a id='toc3_'></a>[Simultaneous solve](#toc0_)

```python
device = choose_gpu()
```

```python
if LOAD_SIMULT:

    model_simult = BufferStockModelClass(load=filename,device=device)

else:

    model_simult = BufferStockModelClass(
        algoname=algoname,device=device,
        par={'Nstates_fixed':0,'N_value_NN':3},
        train={'K_time':K_time_simult})
    
    model_simult.solve(do_print=True,do_print_all=False)
```

```python
print_R(model_simult)
```

## 4. <a id='toc4_'></a>[Backwards refinement](#toc0_)

```python
train = {}
train['backward'] = True
train['use_simult_in_backward'] = True
train['NN_init_std'] = 0.001
train['Nneurons_value_t'] = np.array([50,50])
train['Nneurons_policy_t'] = np.array([50,50])
train['N'] = 1_000_000
train['N_target_batches'] = 10
train['batch_size'] = 100_000
train['K_time'] = K_time
```

```python
algoname = 'DeepVPDBackward'
model_refine = BufferStockModelClass(
    algoname=algoname,device=device,
    par={'Nstates_fixed':0,'N_value_NN':3},train=train
)
```

```python
model_refine.solve(do_print=True,model_simult=model_simult)
print(f'{model_refine.info["time"]/60:.1f} mins')
```

```python
print_R(model_simult)
print_R(model_refine)
```

```python
del model_refine
```

## 5. <a id='toc5_'></a>[Pure backward](#toc0_)

```python
train = {}
train['backward'] = True
train['use_simult_in_backward'] = False
train['Nneurons_value_t'] = np.array([50,50])
train['Nneurons_policy_t'] = np.array([50,50])
train['N'] = 1_000_000
train['N_target_batches'] = 10
train['batch_size'] = 100_000
train['K_time'] = K_time
```

```python
model_pure = BufferStockModelClass(
    algoname=algoname,device=device,
    par={'Nstates_fixed':0,'N_value_NN':3},train=train
)
```

```python
model_pure.solve(do_print=True)
print(f'{model_pure.info["time"]/60:.1f} mins')
```

```python
print_R(model_simult)
print_R(model_pure)
```

```python
del model_pure
```
