## 3. <a id='toc3_'></a>[Run](#toc0_)

```python
def run(algoname,par=None,train=None):

    print(f'### {algoname} ###')
    
    par = {} if par is None else par

    if algoname == 'DeepVPD-FOC': par['use_FOC'] = True  

    train = {} if train is None else train

    algoname_ = algoname if not algoname == 'DeepVPD-FOC' else 'DeepVPD'
    model = BufferStockModelClass(algoname=algoname_,device=device,
                                  par={'Nstates_fixed':0,**par},
                                  train={**train})
    
    return model
```

```python
algonames = ['DeepSimulate','DeepFOC','DeepVPD','DeepVPD-FOC','DeepVPDDC']
algonames += ['DeepFOCBackward','DeepVPDBackward','DeepVPDDCBackward']
models = {}
for algoname in algonames:
    model = run(algoname,par=None,train=None)
    models[algoname] = model
    print('')
```
