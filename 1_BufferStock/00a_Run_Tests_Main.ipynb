```python
device = choose_gpu()
K = 1
```

## 3. <a id='toc3_'></a>[Run](#toc0_)

```python
def run(algoname,par=None,train=None):

    par = {} if par is None else par
    if algoname == 'DeepVPD-FOC': par['use_FOC'] = True  

    train = {} if train is None else train

    print(f'### {algoname} ###')

    algoname_ = algoname if not algoname == 'DeepVPD-FOC' else 'DeepVPD'
    model = BufferStockModelClass(algoname=algoname_,device=device,
                                  par={'Nstates_fixed':0,**par},
                                  train={'K':K,'start_train_policy':-1,**train})
    
    model.solve(do_print=True)  

    return model
```

```python
algonames = ['DeepSimulate','DeepFOC','DeepVPD','DeepVPD-FOC']
models = {}
for algoname in algonames:
    model = run(algoname,par=None,train=None)
    models[algoname] = model
    print('')
```

## 4. <a id='toc4_'></a>[DeepFOC-KKT](#toc0_)

```python
run('DeepFOC',par={'KKT':True})
```

## 5. <a id='toc5_'></a>[MC](#toc0_)

```python
for algoname in algonames:
    if algoname == 'DeepSimulate': continue
    run(algoname,train={'use_quad':False})
    print('')
```

## 6. <a id='toc6_'></a>[No exploration](#toc0_)

```python
for algoname in algonames:
    run(algoname,train={'epsilon_sigma':torch.tensor([0.0])})
    print('')
```

## 7. <a id='toc7_'></a>[Terminal actions](#toc0_)

```python
for algoname in algonames:

    par = {'bequest':0.5}
    train = {'terminal_actions_known':True}

    run(algoname,par=par,train=train)
    print('')
```
