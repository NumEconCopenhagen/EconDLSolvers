## 6. <a id='toc6_'></a>[Manual bias initialization](#toc0_)


In the last layer, we initialize the bias using the inverse of the sigmoid function (also known as "log-odds"):
\begin{align*}
y = f(w_p) &=\frac{1}{1+\exp(-w_p)} \\
\Leftrightarrow w_p = f^{-1}(y) &= \ln \left( \frac{y}{1-y} \right)
\end{align*}
Since all weights are initialized close to zero, this implies that the neural network will output $\approx y$ before any training updates.

```python
par = {}
par['Nstates_fixed'] = 0
par['init_bias_logodds'] = True
par['w_p'] = 0.1

train = {}
train['K_time'] = 0.1
train['manual_init_policy'] = True
```

```python
model_bias_init = BufferStockModelClass(algoname='DeepVPD',device=device,par=par,train=train)
```

Check that network outputs in the neighbourhood of 0.1.

```python
model_bias_init._simulate_training_sample(None)
model_bias_init.train.actions.mean()
```

```python
model_bias_init.solve(do_print=True)
```

## 7. <a id='toc7_'></a>[Manual weight initialization](#toc0_)

In the following we consider Xavier and He initialization of respectively sigmoid and ReLU layers. 

Both Xavier and He initialization can be drawn using the uniform or the normal distribution.

He uniform: $ \qquad \mathbf{w}_{i,j} = \mathcal{U} \left( -\sqrt{\frac{6}{\text{\#neurons}_i}}, \sqrt{\frac{6}{\text{\#neurons}_i}} \right) $

He normal: $ \qquad \mathbf{w}_{i,j} = \mathcal{N} \left(0, \frac{2}{\text{\#neurons}_i} \right) $

Xavier uniform: $ \qquad \mathbf{w}_{i,j} = \mathcal{U} \left( -\sqrt{\frac{6}{\text{\#neurons}_i+\text{\#neurons}_j}}, \sqrt{\frac{6}{\text{\#neurons}_i+\text{\#neurons}_j}} \right) $

Xavier normal: $ \qquad \mathbf{w}_{i,j} = \mathcal{N} \left(0, \frac{2}{\text{\#neurons}_i+\text{\#neurons}_j} \right) $

```python
par = {}
par['Nstates_fixed'] = 0
par['init_sigmoid_xavier_uniform'] = False # initialize sigmoid layers with xavier uniform
par['init_sigmoid_xavier_normal'] = False # initialize sigmoid layers with xavier normal
par['init_relu_he_normal'] = False # initialize relu layers with he normal
par['init_relu_he_uniform'] = True # initialize relu layers with he uniform

train = {}
train['K_time'] = 0.1
train['manual_init_policy'] = True
```

```python
model_w_init = BufferStockModelClass(algoname=algoname,device=device,par=par,train=train)
```

```python
model_w_init.solve(do_print=True)
```

```python
del model_w_init
```
