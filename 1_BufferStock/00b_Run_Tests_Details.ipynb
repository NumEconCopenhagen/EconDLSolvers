## 2. <a id='toc2_'></a>[Setup](#toc0_)

```python
algoname = 'DeepVPD'
```

```python
device = choose_gpu()
```

```python
model = BufferStockModelClass(algoname=algoname,device=device,
                              par={'Nstates_fixed':0,'full':False},
                              train={'K':51,'sim_R_freq':10,'N_value_NN':3})
```

## 3. <a id='toc3_'></a>[Run](#toc0_)

```python
model.solve(do_print=True,do_print_all=False)
```

**Note:** Manually stop by setting `true` in `solving.json` next to the timestamp. 

**Print detailed timings:**

```python
for k,v in model.info.items():
    if type(k) == str and k[:4] == 'time':
        print(f'{k}: {v:.1f} secs')  
```

**Memory:**

```python
if torch.cuda.is_available():
    print(f'CUDA memory allocated (GB): {torch.cuda.max_memory_allocated(device)/10**9}')
```

**Save:**

```python
model.save('test.pt')
```

### 3.1. <a id='toc3_1_'></a>[Continue](#toc0_)

### 3.3. <a id='toc3_3_'></a>[Same](#toc0_)

```python
model_same = BufferStockModelClass(algoname=algoname,device=device,
                                   par={'Nstates_fixed':0,'full':False},
                                   train={'K':model.train.sim_R_freq+1})
model_same.solve(do_print=True)
del model_same
```

### 3.4. <a id='toc3_4_'></a>[New](#toc0_)

1. New sample for lifetime reward
1. New initial values for neural nets
1. New training samples

```python
torch_rng_state = torch.get_rng_state()
model_new = BufferStockModelClass(algoname=algoname,device=device,
                                  par={'Nstates_fixed':0,'full':False},
                                  train={'K':model.train.sim_R_freq+1},
                                  torch_rng_state=torch_rng_state)
model_new.solve(do_print=True)
del model_new
```

## 4. <a id='toc4_'></a>[Timing](#toc0_)

```python
algonames = ['DeepSimulate', 'DeepFOC']
for algoname in algonames:
    model_time = BufferStockModelClass(algoname=algoname,device=device,par={'Nstates_fixed':0},train={'K_time':10})
    model_time.time_solve()
```
